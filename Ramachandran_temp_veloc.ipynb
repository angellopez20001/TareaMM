```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm


# 1. Cargar datos
data = np.loadtxt("ALA-2.dat")  

phi = data[:, 1]   # columna φ
psi = data[:, 2]   # columna ψ

#2. Ramachandran plot con densidad 2D 
fig, ax = plt.subplots(figsize=(6, 6))


counts, xedges, yedges = np.histogram2d(
    phi, psi,
    bins=100,
    range=[[-180, 180], [-180, 180]],
    density=True   
)


counts_masked = np.ma.masked_where(counts == 0, counts)


img = ax.imshow(
    counts_masked.T,
    origin="lower",
    extent=[-180, 180, -180, 180],
    aspect="equal",
    cmap="YlGn",          
    vmin=0, vmax=0.09,    
    interpolation="bilinear"
)


cbar = plt.colorbar(img, ax=ax)
cbar.set_ticks(np.arange(0, 0.10, 0.01))


ax.set_xlim(-180, 180)
ax.set_ylim(-180, 180)
ax.set_xlabel("φ", fontsize=13)
ax.set_ylabel("ψ", fontsize=13)
ax.set_xticks([-180, -90, 0, 90, 180])
ax.set_yticks([-180, -90, 0, 90, 180])
ax.set_title("raman-ALA-2.dat\ncharmm36", fontsize=9)
ax.axhline(0, color="gray", linewidth=0.5, linestyle="--")
ax.axvline(0, color="gray", linewidth=0.5, linestyle="--")

plt.tight_layout()
plt.savefig("ramachandranALA-2.png", dpi=150)
plt.show()

# 3. Identificar conformaciones de alta probabilidad 

regiones = {
    "alpha_R": {"phi": (-160, -20),  "psi": (-120, 60)},
    "beta":    {"phi": (-180, -90),  "psi": (90,  180)},
    "alpha_L": {"phi": (20,   80),   "psi": (20,   80)},
}

print("\nPorcentaje de población en cada conformación:")
n_total = len(phi)

for nombre, rango in regiones.items():
    mask = (
        (phi >= rango["phi"][0]) & (phi <= rango["phi"][1]) &
        (psi >= rango["psi"][0]) & (psi <= rango["psi"][1])
    )
    pct = mask.sum() / n_total * 100
    print(f"  {nombre:10s}: {pct:.1f}%")
```

```python
def load_xvg(filename):
    data = []
    with open(filename, 'r') as f:
        for line in f:
            if line.startswith('#') or line.startswith('@'):
                continue
            data.append(line.split())
    return np.array(data, dtype=float)

temp_data  = load_xvg("temp.xvg")
veloc_data = load_xvg("veloc.xvg")

time_temp  = temp_data[:, 0]   # tiempo (ps)
temp       = temp_data[:, 1]   # temperatura (K)

time_vel   = veloc_data[:, 0]  # tiempo (ps)
vx         = veloc_data[:, 1]  # componente x
vy         = veloc_data[:, 2]  # componente y
vz         = veloc_data[:, 3]  # componente z

# ── 2. Plot ──────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
fig.suptitle("Histogramas de Temperatura y Velocidad", fontsize=13)

# Configuración común
bins = 50
color_temp = "tomato"
colores_vel = ["steelblue", "seagreen", "darkorchid"]

# ── Temperatura ──────────────────────────────────────────────────────────────
axes[0, 0].hist(temp, bins=bins, color=color_temp, edgecolor="white")
axes[0, 0].set_xlabel("Temperatura (K)", fontsize=11)
axes[0, 0].set_ylabel("Frecuencia", fontsize=10)
axes[0, 0].set_title("Temperatura")
axes[0, 0].axvline(np.mean(temp), color="black", linestyle="--", 
                    linewidth=1.2, label=f"Media = {np.mean(temp):.1f} K")
axes[0, 0].legend(fontsize=9)

# ── Velocidad Vx ─────────────────────────────────────────────────────────────
axes[0, 1].hist(vx, bins=bins, color=colores_vel[0], edgecolor="white")
axes[0, 1].set_xlabel("Vx (nm/ps)", fontsize=11)
axes[0, 1].set_ylabel("Frecuencia", fontsize=10)
axes[0, 1].set_title("Velocidad — componente X")
axes[0, 1].axvline(0, color="black", linestyle="--", linewidth=1.0)

# ── Velocidad Vy ─────────────────────────────────────────────────────────────
axes[1, 0].hist(vy, bins=bins, color=colores_vel[1], edgecolor="white")
axes[1, 0].set_xlabel("Vy (nm/ps)", fontsize=11)
axes[1, 0].set_ylabel("Frecuencia", fontsize=10)
axes[1, 0].set_title("Velocidad — componente Y")
axes[1, 0].axvline(0, color="black", linestyle="--", linewidth=1.0)

# ── Velocidad Vz ─────────────────────────────────────────────────────────────
axes[1, 1].hist(vz, bins=bins, color=colores_vel[2], edgecolor="white")
axes[1, 1].set_xlabel("Vz (nm/ps)", fontsize=11)
axes[1, 1].set_ylabel("Frecuencia", fontsize=10)
axes[1, 1].set_title("Velocidad — componente Z")
axes[1, 1].axvline(0, color="black", linestyle="--", linewidth=1.0)

plt.tight_layout()
plt.savefig("histogramas.png", dpi=150, bbox_inches="tight")
plt.show()

# ── 3. Estadísticas básicas ──────────────────────────────────────────────────
print(f"Temperatura  → Media: {np.mean(temp):.2f} K  |  Std: {np.std(temp):.2f} K")
print(f"Vx           → Media: {np.mean(vx):.4f}  |  Std: {np.std(vx):.4f} nm/ps")
print(f"Vy           → Media: {np.mean(vy):.4f}  |  Std: {np.std(vy):.4f} nm/ps")
print(f"Vz           → Media: {np.mean(vz):.4f}  |  Std: {np.std(vz):.4f} nm/ps")
```
